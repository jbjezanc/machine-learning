# Randomly generated arrays

```python
# Uniform distribution between 0 and 1
np.random.rand(5, 2)
```

```python
# define a random seed to always return reproducible random array (pseudo rundom numbers)
np.random.seed(2)
100 * np.random.rand(5, 2)
```

```python
# reproducible sample in standard normal distribution
np.random.seed(2)
np.random.randn(5, 2)
```

```python
# reproducible matrix of pseudorandom integers between 0 and 100
np.random.seed(2)
np.random.randint(low=0, high=100, size=(5, 2))
```
